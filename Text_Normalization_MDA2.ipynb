**Lowercasing**

All alphabetic characters in the tweets are converted to lowercase to maintain a consistent text format.

```python
df_sample["lowercase_text"] = df_sample["text"].str.lower()

print("Before and After Lowercasing:")
display(df_sample[["text", "lowercase_text"]].head(10))
```

```python
contractions = {
    "don't": "do not",
    "didn't": "did not",
    "it's": "it is",
    "isn't": "is not",
    "can't": "cannot",
    "won't": "will not",
    "you've": "you have",
    "I've": "I have",
    "I'm": "I am",
    "we're": "we are",
    "they're": "they are"
}

def expand_contractions(text):
    for contraction, expanded in contractions.items():
        text = re.sub(
            r"\b" + re.escape(contraction) + r"\b",
            expanded,
            text,
            flags=re.IGNORECASE
        )
    return text

df_sample["expanded_text"] = df_sample["lowercase_text"].apply(expand_contractions)

print("Before and After Contraction Expansion:")
display(df_sample[["lowercase_text", "expanded_text"]].head(10))
```

**Contraction Expansion**
Common contractions in the tweets are expanded into their complete forms.
Examples:
- don't → do not
- didn't → did not


**Punctuation Handling**
Punctuation marks are removed from the lowercase text to create a simpler and more consistent text format.

```python
df_sample["no_punctuation_text"] = df_sample["lowercase_text"].apply(
    lambda x: re.sub(r"[^\w\s]", "", x)
)

print("Before and After Punctuation Handling:")
display(df_sample[["lowercase_text", "no_punctuation_text"]].head(10))
```

**Note:** In the final normalization, contractions are expanded before punctuation is removed so that forms such as "didn't" can be correctly converted to "did not".

**Handle Numbers and Emojis**
Numbers and emojis are removed using simple regular expression rules to obtain cleaner text.

example:"I paid $30 😊" → "I paid"

```python
def handle_numbers_emojis(text):
    text = re.sub(r"\d+", "", text)
    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

df_sample["normalized_text"] = df_sample["expanded_text"].apply(
    handle_numbers_emojis
)

print("Before and After Handling Numbers and Emojis:")
display(df_sample[["expanded_text", "normalized_text"]].head(10))
```

**Compare Before and After Normalization**

The original tweets are compared with the normalized tweets to observe the changes made during the normalization process.

```python
comparison = df_sample[["text", "normalized_text"]].head(10).copy()

comparison.columns = ["Original Text", "Normalized Text"]

display(comparison)
```
